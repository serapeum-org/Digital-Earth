# The cleopatra preset library, through Digital-Earth

cleopatra 0.31 ships a library of ~200 named **parameter presets** (`cleopatra.styling.colors.DATA_STYLES`). Each preset bundles a colormap, a normalization, and a human label tuned for a specific geophysical variable — elevation, sea-surface temperature, radar reflectivity, aerosol optical depth, and so on.

Digital-Earth exposes the whole library through **one keyword**: pass `style="<preset>"` to any raster method (`field`, `contours`, `pcolormesh`, `grid_cells`, …) and the map picks up that preset's colormap and label. This notebook tours the library; the siblings drill into domains, palettes, and colour scaling.

```python
%matplotlib inline
from pathlib import Path
import numpy as np
from pyramids.dataset import Dataset
from digitalearth.static import Map, grid
from cleopatra.styling.colors import DATA_STYLES

DATA = Path('../../../examples/data')   # notebook-relative path (docs/examples/presets -> repo root)
dem = Dataset.read_file(str(DATA / 'LisbonElevation.tif'))   # a small bundled DEM (EPSG:4326)
dem
```

## What is in the library

Every preset is a named entry cleopatra resolves from `DATA_STYLES`. They span terrain, weather, ocean, atmospheric-composition and radar families, plus the full set of perceptually-uniform *scientific* colour maps (batlow, broc, hawaii, davos, …).

```python
presets = sorted({sub for v in DATA_STYLES.values() if isinstance(v, dict) for sub in v})
print(len(presets), 'presets available')
print('examples:', presets[:16])
```

## One field, many presets

The quickest way to feel the library is to render the **same** elevation field under a spread of presets. `digitalearth.static.grid` lays out a grid of shared-CRS maps; each panel just calls `field(dem, style=...)`.

```python
show = ['elevation', 'bathymetry', 'batlow', 'hawaii', 'broc',
        'davos', 'grayC', 'temperature', 'anomaly']
fig, maps = grid(3, 3, crs=dem.epsg, figsize=(13, 11))
for m, preset in zip(maps, show):
    m.field(dem, style=preset)
    m.set_title(preset)
fig;
```

## A preset carries its label

A preset is more than a colormap — it also knows what it is plotting. Rendering one on its own and adding a colorbar shows the preset's colormap and its label together.

```python
m = Map(crs=dem.epsg, figsize=(9, 7))
m.field(dem, style='elevation')
m.colorbar(label='elevation (m)')
m.set_title("style='elevation'")
m.fig;
```
