# 7 · One-call API — quickplot / quickmap

`digitalearth.quickplot` builds a finished, decorated `Map` in one call, dispatching on input type. Module functions (`imshow`, `contourf`, `scatter`, `choropleth`, ...) mirror the methods.

```python
%matplotlib inline
from pathlib import Path

# Resolve the repo root so the bundled sample data is found whether this runs
# from docs/examples/ (mkdocs) or the repository root.
ROOT = Path.cwd()
while not (ROOT / "examples" / "data" / "acc4000.tif").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
RASTER = str(ROOT / "examples" / "data" / "acc4000.tif")
POINTS = str(ROOT / "tests" / "data" / "points.geojson")
print("data root:", ROOT)
```

```python
import digitalearth
from digitalearth import api
from pyramids.dataset import Dataset
from pyramids.feature import FeatureCollection

ds = Dataset.read_file(RASTER)
```

## A raster in one call (auto-styled, with a colorbar)

```python
m = digitalearth.quickplot(ds, crs=ds.epsg)
m.set_title('quickplot(dataset)')
m.fig
```

## Choose the renderer

```python
m = api.contourf(ds, crs=ds.epsg, levels=10)
m.set_title('api.contourf')
m.fig
```

## Points dispatch to a scatter map

```python
fc = FeatureCollection.read_file(POINTS)
m = digitalearth.quickmap(fc, crs=fc.epsg)
m.set_title('quickmap(points)')
m.fig
```
