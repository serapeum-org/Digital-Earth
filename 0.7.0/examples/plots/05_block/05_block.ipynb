# 5 · `block`

`Map.block` renders a raster as **discrete cell blocks** (cleopatra `ArrayGlyph`, `kind="block"`). `Map` reprojects via pyramids and draws on a shared axes; `colorbar()` adds one aggregated bar.

**Setup** — inline plots and the path to the bundled Lisbon DEM.

```python
%matplotlib inline
from pathlib import Path

# Locate the bundled Lisbon DEM from the repo root (works from docs/ or the repo root).
ROOT = Path.cwd()
while not (ROOT / "examples" / "data" / "LisbonElevation.tif").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
DEM = str(ROOT / "examples" / "data" / "LisbonElevation.tif")
from pyramids.dataset import Dataset
from digitalearth.scene import Map
ds = Dataset.read_file(DEM)
```

Draw the Lisbon DEM with `Map.block`. `pcolormesh` on cell edges.

```python
m = Map(crs=ds.epsg, figsize=(9, 7))
m.block(ds, cmap='terrain')
m.colorbar(label='elevation (m)')
m.set_title('block')
m.fig;
```
