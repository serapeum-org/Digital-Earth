# 22 · `boxplot`

`series.boxplot` draws a box-and-whisker of a 1-D sample (cleopatra `HistogramGlyph`).

**Setup** — inline plots and the path to the bundled Lisbon DEM.

```python
%matplotlib inline
from pathlib import Path

# Locate the bundled Lisbon DEM from the repo root (works from docs/ or the repo root).
ROOT = Path.cwd()
while not (ROOT / "examples" / "data" / "LisbonElevation.tif").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
DEM = str(ROOT / "examples" / "data" / "LisbonElevation.tif")
import numpy as np
import matplotlib.pyplot as plt
from digitalearth import series
rng = np.random.default_rng(0)
```

Box-and-whisker of a sample.

```python
fig, ax = plt.subplots(figsize=(5, 4))
series.boxplot(rng.normal(0, 1, 500), ax=ax)
ax.set_title('boxplot')
fig;
```
