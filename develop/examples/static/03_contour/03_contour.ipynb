# 3 · `contours`

`Map.contours` renders a raster as **line contours (isolines)** (cleopatra `ArrayGlyph`, `kind="contour"`). `Map` reprojects via pyramids and draws on a shared axes; `colorbar()` adds one aggregated bar.

**Setup** — inline plots and the path to the bundled Lisbon DEM.

```python
%matplotlib inline
from pathlib import Path

# Locate the bundled Lisbon DEM from the repo root (works from docs/ or the repo root).
ROOT = Path.cwd()
while not (ROOT / "examples" / "data" / "LisbonElevation.tif").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
DEM = str(ROOT / "examples" / "data" / "LisbonElevation.tif")
from pyramids.dataset import Dataset
from digitalearth.static import Map
ds = Dataset.read_file(DEM)
```

Draw the Lisbon DEM with `Map.contours`. No fill, so no colorbar.

```python
m = Map(crs=ds.epsg, figsize=(9, 7))
m.contours(ds, cmap='terrain')
m.set_title('contour')
m.fig;
```
